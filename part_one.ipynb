### Sous jacent

#### 1. Base 100 du spot (XS1732593444)

```python
def base100(series):
    return (series / series.iloc[0]) * 100

data["Base 100"] = base100(data["XS1732593444"])

data
```

#### 2. Performances

```python
# Daily

def daily_return(series):
    return series.pct_change()

data["Daily Return"] = daily_return(data["Base 100"])

# Cumulée
def cumulative_return(series):
    return series / series.iloc[0] - 1

data["Cumulative Return"] = cumulative_return(data["Base 100"])

data
```

#### 3. Drawdown Index

```python
def drawdown(series):
    return (series / series.cummax()) - 1

data["Drawdown"] = drawdown(data["Base 100"])

data
```

#### 4. Max DD

```python
def max_drawdown(series):
    return series.cummin()

data["Maximum Drawdown"] = max_drawdown(data["Drawdown"])

data
```

#### 5. Volatility ITD

```python
def volatility_itd(series):
    return series.pct_change().expanding().std(ddof=1) * np.sqrt(252)

data['Volatility'] = volatility_itd(data["Base 100"])

data
```

#### 5. Minimum / Maximum Level Spot and Base 100

```python
def min_level(series):
    return series.cummin()

def max_level(series):
    return series.cummax()

data["Minimum Spot Level"] = min_level(data["XS1732593444"])
data["Maximum Spot Level"] = max_level(data["XS1732593444"])

data["Minimum Base 100 Level"] = min_level(data["Base 100"])
data["Maximum Base 100 Level"] = max_level(data["Base 100"])

data
```

### Autocall

#### 1. Autocall Level : Niveau de remboursement anticipé automatique

```python
def autocall_level(series, level):
    initial_value = series.iloc[0]
    return initial_value * level

data["Autocall Level"] = autocall_level(data["XS1732593444"], 1)

data
```

#### 2. Autocall Distance : Marge avant déclenchement de l'autocall

```python
def autocall_distance(series, series_autocall_level):
    return (series / series_autocall_level) - 1

data["Distance to Autocall"] = autocall_distance(data["XS1732593444"], data["Autocall Level"])

data
```

#### 3. Next Observation Date : Prochaine date d'observation

```python
def next_observation_date(data_series, calendar_series):

    # Recherche de la première date du calendrier >= à chaque date de data_series
    indices = calendar_series.searchsorted(data_series, side="left")

    # Création de la série résultat
    result = pd.Series(pd.NaT, index=data_series.index, dtype="datetime64[ns]")

    # On ne garde que les indices valides
    mask = indices < len(calendar_series)

    # Attribution de la prochaine date d'observation
    result.loc[mask] = calendar_series.iloc[indices[mask]].values

    return result

data["Next Observation Date"] = next_observation_date(data["Date"], calendar["Evaluation Date"])

data
```

#### 4. n = numéro de la prochaine Next Observation Date

```python
def next_observation_number(data_series, calendar):
    
    # Recherche de la première Evaluation Date >= à chaque date de data_series
    indices = calendar["Evaluation Date"].searchsorted(data_series, side="left")

    # Création de la série résultat
    result = pd.Series(pd.NA, index=data_series.index, dtype="Int64")

    # On ne garde que les indices valides
    mask = indices < len(calendar)

    # Attribution du n correspondant à la prochaine date d'observation
    result.loc[mask] = calendar["n"].iloc[indices[mask]].values

    return result

data["n of Next Observation Date"] = next_observation_number(data["Date"], calendar)

data
```

#### 5. Days to Observation

```python
def nb_days_to_next_observation(date_series, next_obs_series):
    # Calcul du nombre de jours entre la date de data et la prochaine date d'observation
    return (next_obs_series - date_series).dt.days

data["Days to Next Observation"] = nb_days_to_next_observation(data["Date"], data["Next Observation Date"])

data
```

#### 6. Potential Redemption Amount : Montant payé si autocall à la prochaine constatation rappelé

```python
def potential_redemption_amount(observation_number_series):
    return (1000 * (1.035 + observation_number_series * 0.035)).round(0)

data["Potential Redemption Amount"] = potential_redemption_amount(data["n of Next Observation Date"])

data
```

#### 7. Potential Coupon / Premium : Gain accumulé conditionnel

```python
def potential_coupon_prenium(observation_number_series):
    return (1000 * 0.035 * (observation_number_series + 1)).round(0)

data["Potential Coupon Premium"] = potential_coupon_prenium(data["n of Next Observation Date"])

data
```

#### 8. Potential Product Performance

```python
def potential_redemption_amount(D, series_potential_redemption):
    return (series_potential_redemption / D) - 1

data["Potential Product Performance"] = potential_redemption_amount(1000, data["Potential Redemption Amount"])

data
```

### Protection

#### 1. Capital Barrier : Niveau de la barrière de protection du capital

```python
def capital_barrier(series, barrier):
    initial_value = series.iloc[0]
    return initial_value * barrier

data["Capital Barrier"] = capital_barrier(data["XS1732593444"], 0.6)

data
```

#### 2. Distance to Capital Barrier

```python
def barrier_distance(series, series_barrier):
    return (series / series_barrier) - 1

data["Distance to Capital Barrier"] = barrier_distance(data["XS1732593444"], data["Capital Barrier"])

data
```

#### 3. Final Protection Level of Nominal : Remboursement nominal si 0.6 <= ST <= 0.8

```python
data["Final Protection Amount"] = data["Capital Barrier"].copy()

data
```

#### 4. Threshold Barrier : barrière en dessous de laquelle dans la limite de la capital barrier on ne rembourse que le nominal

```python
def threshold_barrier(series, threshold_level):
    initial_value = series.iloc[0]
    return initial_value * threshold_level

data["Threshold Barrier"] = threshold_barrier(data["XS1732593444"], 0.8)

data
```

#### 5. Distance to Threshold

```python
def barrier_distance(series, series_barrier):
    return (series / series_barrier) - 1

data["Distance to Threshold Barrier"] = barrier_distance(data["XS1732593444"], data["Threshold Barrier"])

data
```
